```python

import numpy as np
import scipy.stats as stats


def get_ground_truth(query,dataframe):
    city_list = [city[0] for city in query]
    ground_truth1_1 = dataframe.iloc[dataframe['city'].isin(city_list)]

    return ground_truth1_1

def precision_at_three(ground_truth, results):
    top_three_results = [result[0] for result in results[:3]]
    
    relevant_cities = ground_truth.head(3)['city'].to_list()
    total = sum(1 for city in top_three_results if city in relevant_cities)
    print(total/3)
    
    return total/3

def kendal_tau(ground_truth, results):
    ground_truth_ranking = {city: rank for rank, city in enumerate(ground_truth['city'].to_list())}
    results_ranking = {city: rank for rank, (city, _) in enumerate(results)}
    
    tau, p_value = stats.kendalltau(
        [ground_truth_ranking[city] for city in ground_truth['city'].to_list()],
        [results_ranking[city] for city in ground_truth['city'].to_list()]
    )
    return tau, p_value

def merge_kendall_tau_stat(list_tau, list_p_value):
    
    
    tau_array = np.array(list_tau)
    p_array = np.array(list_p_value)
    
    avg_tau = np.mean(tau_array)
    
    statistics, global_p = stats.combine_pvalues(p_array, method='fisher')
    
    # 3. Confezioniamo il risultato con una chiave di lettura chiara
    results = {
        "avg_tau": avg_tau,
        "global_p": global_p,
        "significance": global_p <= 0.05
    }
    
    return results

```
